# 04 · Held-out model evaluation

Evaluate an actually trained, checksum-verified model through `covid_xray.evaluate`. This notebook contains no precomputed metrics. Run it once after model selection; repeated tuning against test results invalidates the holdout.

```python
from pathlib import Path
import json
import os
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
os.chdir(ROOT)
MANIFEST = ROOT / 'data' / 'processed' / 'manifest.csv'
MODEL = ROOT / 'artifacts' / 'models' / 'model.keras'
RESULTS = ROOT / 'artifacts' / 'results'
missing = [str(path) for path in (MANIFEST, MODEL) if not path.is_file()]
if missing:
    raise FileNotFoundError('Complete preprocessing and training first; missing: ' + ', '.join(missing))
```

```python
subprocess.run([sys.executable, '-m', 'covid_xray.evaluate', '--manifest', str(MANIFEST), '--model', str(MODEL), '--output', str(RESULTS)], check=True)
```

```python
metrics = json.loads((RESULTS / 'metrics.json').read_text())
display(metrics)
print(f'Review confusion matrix, ROC curves and error analysis under: {RESULTS}')
print('These experimental image-level results are not medical diagnoses.')
```
