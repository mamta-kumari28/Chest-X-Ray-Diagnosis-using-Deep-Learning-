# 03 · Custom CNN training

Train through the existing `covid_xray.train` module. This notebook contains no saved output or fabricated result. Use a Colab GPU only after preprocessing the licensed dataset. The test split is not passed to model fitting.

```python
from pathlib import Path
import os
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
os.chdir(ROOT)
MANIFEST = ROOT / 'data' / 'processed' / 'manifest.csv'
MODELS = ROOT / 'artifacts' / 'models'
if not MANIFEST.is_file():
    raise FileNotFoundError('Run 02_preprocessing.ipynb first.')
```

```python
import tensorflow as tf

print('TensorFlow:', tf.__version__)
gpus = tf.config.list_physical_devices('GPU')
print('GPUs:', gpus)
assert gpus, 'Select a GPU runtime before full real-data training.'
```

```python
EPOCHS = 30
BATCH_SIZE = 32
subprocess.run([sys.executable, '-m', 'covid_xray.train', '--manifest', str(MANIFEST), '--output', str(MODELS), '--epochs', str(EPOCHS), '--batch-size', str(BATCH_SIZE)], check=True)
print(f'Best-model bundle: {MODELS}')
```
